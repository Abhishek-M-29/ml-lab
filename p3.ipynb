```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA
```

Lead Iris dataset

```python
iris = load_iris()
data = iris.data
labels = iris.target
label_names = iris.target_names
```

Convert to a DataFrame for better visualization

```python
iris_df = pd.DataFrame(data, columns=iris.feature_names)
```

Perform PCA to reduce dimentionality to 2

```python
pca = PCA(n_components=2)
data_reduced = pca.fit_transform(data)
```

Create a DataFrame for reduced data

```python
reduced_df = pd.DataFrame(data_reduced, columns=['Principal Component 1', 'Principal Component 2'])
reduced_df['Label'] = labels
```

Plot the Reduced data

```python
plt.figure(figsize=(8, 6))
colors = ['r', 'g', 'b']
for i, target_name in enumerate(label_names):
    plt.scatter(
        reduced_df[reduced_df['Label'] == i]['Principal Component 1'],
        reduced_df[reduced_df['Label'] == i]['Principal Component 2'],
        label=target_name,
        color = colors[i]
    )
plt.title('PCA of Iris Dataset')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend()
plt.grid()
plt.show()
```
